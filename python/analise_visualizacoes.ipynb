```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
arquivo_dataset = '../dados/analise_logistica_dataset.xlsx'

pedidos = pd.read_excel(arquivo_dataset, sheet_name='PEDIDOS')
produtos = pd.read_excel(arquivo_dataset, sheet_name='PRODUTOS')
```

```python
vendas_por_produto = (
    pedidos.merge(
        produtos[['produto_id', 'produto_nome']],
        on='produto_id'
    )
    .groupby('produto_nome')['quantidade']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)
```

## Visualização de Top 5 Produtos mais Vendidos

```python
plt.figure(figsize=(9, 9))
vendas_por_produto.plot(
    kind='bar',
    color=['green', 'orange', 'orange', 'orange', 'red'],
    width=0.75)

plt.title('Top 5 Produtos mais Vendidos')
plt.xlabel('Produto')
plt.ylabel("Quantidade de Unidades Vendidas")
plt.xticks(rotation=50, ha='right')
plt.tight_layout()

plt.show()
```

```python
clientes = pd.read_excel(arquivo_dataset, sheet_name='CLIENTES')
```

```python
faturamento_por_regiao = (
    pedidos.merge(
        clientes[['cliente_id', 'regiao']],
        on='cliente_id'
    )
    .groupby('regiao')['valor_total']
    .sum()
    .sort_values(ascending=False)
)
```

## Visualização de Faturamento por Região

```python
plt.figure(figsize=(8, 5))

faturamento_por_regiao.plot(
    kind='bar',
    color=['green', 'orange', 'orange', 'orange', 'red'],
    width=0.75)

plt.title('Faturamento por Região')
plt.xlabel('Região')
plt.ylabel('Faturamento (R$)')
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
entregas = pd.read_excel(
    arquivo_dataset,
    sheet_name='ENTREGAS'
)
```

```python
status_entregas = entregas['status_entrega'].value_counts()
```

## Visualização de Status de Entrega

```python
plt.figure(figsize=(8, 5))

status_entregas.plot(
    kind='bar',
    color=['green', 'red'],
    width=0.7)

plt.title('Status das Entregas')
plt.xlabel('Status')
plt.ylabel('Quantidade de Entregas')
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```
