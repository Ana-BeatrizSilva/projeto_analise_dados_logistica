```python
vendas_produtos = pedidos.merge(
    produtos[['produto_id', 'produto_nome']],
    on='produto_id'
)

vendas_produtos.groupby('produto_nome')['quantidade'].sum().sort_values(ascending=False)
```

```python
vendas_por_produto = (
    vendas_produtos
    .groupby('produto_nome')['quantidade']
    .sum()
    .sort_values(ascending=False)
)
```

## Principais Insights

- O faturamento total dos 100 pedidos foi de **R$ 197.670,00**, com um ticket médio de **R$ 1.976,70 por pedido**.

- O **Pen Drive 128GB** foi o produto com maior quantidade de unidades vendidas, totalizando **42 unidades**.

- A região **Sudeste** apresentou o maior faturamento, com **R$ 59.850,00**, enquanto o **Nordeste** apresentou o menor, com **R$ 26.300,00**.

- Do total de 100 entregas analisadas, **69 foram entregues** e **31 foram classificadas como atrasadas**, representando **31% de entregas atrasadas**.

- Os produtos com maior quantidade vendida não são necessariamente os que apresentam maior faturamento, devido às diferenças de preço entre os produtos.
